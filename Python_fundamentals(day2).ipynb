### Explanation for Q3.3: Integer Caching in Python

Python has an optimization called **integer caching (or interning)** for small integers. To save memory and improve performance, Python pre-allocates and caches a range of commonly used integers, typically from -5 to 256.

When you create an integer within this range, Python reuses the existing object in memory instead of creating a new one. This means that if you assign `x = 100` and `y = 100`, both `x` and `y` will refer to the *exact same object* in memory (because 100 is within the cached range). Therefore, `x is y` evaluates to `True`.

For integers outside this cached range (like 1000), Python creates new distinct objects for each assignment. So, when you assign `p = 1000` and `q = 1000`, `p` and `q` refer to two different objects in memory that just happen to have the same value. Consequently, `p is q` evaluates to `False`, while `p == q` (checking for value equality) remains `True`.

Q(Doubt): But why this feature was created and then this law only is valid upto 256?


The integer caching optimization in Python (for integers typically from -5 to 256) exists primarily for two reasons:

Memory Efficiency: Small integers are used very frequently in Python programs. If Python were to create a new object in memory every time a small integer was used, it would consume a significant amount of memory. By pre-allocating and reusing these common integer objects, Python reduces memory overhead.

Performance: Creating new objects and managing memory takes time. By caching frequently used small integers, Python avoids these overheads, leading to faster execution for operations involving these numbers.

The specific range of -5 to 256 is chosen because these integers are statistically the most commonly used in typical programming scenarios. It provides a good balance between the benefits of caching and the memory cost of pre-allocating a larger range of numbers. Larger integers are less frequently used, so the benefits of caching them would not outweigh the increased memory consumption of storing many more pre-allocated objects.

Q8.3. Now write a function increment_counter() that tries to do counter = counter + 1 without using the global keyword. Run it and explain the error you get.

```python
counter = 0

def increment_counter():
    counter = counter + 1
    print(counter)

increment_counter()
```

When Python sees an assignment statement (counter = ...) inside a function, it automatically treats counter as a local variable for that function. When executing counter + 1, Python looks for a local variable named counter on the right-hand side before assignment, finds nothing, and raises an UnboundLocalError instead of falling back to the global variable.

Q8.4. Fix the function from the previous question using the global keyword so it correctly increments the global counter variable.

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1
    print(counter)

increment_counter()
print(counter)     #(global variable is successfully updated)
```
