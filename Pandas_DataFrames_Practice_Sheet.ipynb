Q-28. Create a DataFrame named school_df from a dictionary containing Student, Class, and Section. Q-29. Create a DataFrame named inventory_df from equal-length lists. Add a new column using a partially indexed Series so that values appear only for selected row indexes. Display the result, then delete the added column and display the DataFrame again.

```python
inventory_df = pd.DataFrame({
    'Item': ['Laptop', 'Tablet', 'Phone'],
    'Category': ['Electronics', 'Electronics', 'Electronics'],
    'Price': [1000,2000,3000]})
inventory_df['Quantity'] = pd.Series([10, 20, 15], index=[0, 2, 4])
print(inventory_df)
print()
del inventory_df['Quantity']
print(inventory_df)
```
